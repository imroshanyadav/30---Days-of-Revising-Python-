```python
# Different python data types
# Let's declare variables with various data types

first_name = 'Asabeneh'     # str
last_name = 'Yetayeh'       # str
country = 'Finland'         # str
city= 'Helsinki'            # str
age = 250                   # int, it is not my real age, don't worry about it

# Printing out types
print(type('Asabeneh'))          # str
print(type(first_name))          # str
print(type(10))                  # int
print(type(3.14))                # float
print(type(1 + 1j))              # complex
print(type(True))                # bool
print(type([1, 2, 3, 4]))        # list
print(type({'name':'Asabeneh'})) # dict
print(type((1,2)))               # tuple
print(type(zip([1,2],[3,4])))    # zip
```

```python
# Type Casting (Converting the one datatype to other datatype)

#int to float
num = 12
print(type(num))
print(num)
num =float(num)
print(type(num))
print(num)

#float to int

num2 =43.2
print(type(num2))
print(num2)
num2 = int(num2)
print(type(int(num2)))
print(num2)

#int to str

num3 = 43
print(type(num3))
print(num3)
print(type(str(num3)))
print(num3)

# str to int to float

num4 = "43"
print(type(num4))
print(num4)
print(type(int(num4)))
print(num4)
print(type(float(num4)))
print(num4)

# string to list

num5 = "Hello My name is Roshan Yadav"
print(type(num5))
print(num5)
num5 =list(num5)
print(type(list(num5)))
print(num5)
```
