```python
# some escape sequence charectors
print("hello\nworld") # new line

print("hello\tworld")# gap
print("hello\\world")# back slash
print("hello\'world")# double or single 
print("hello\"world")
```

```python
# Strings only
first_name = 'Asabeneh'
last_name = 'Yetayeh'
language = 'Python'
formated_string = 'I am %s %s. I teach %s' %(first_name, last_name, language)
print(formated_string)

# Strings  and numbers
radius = 10
pi = 3.14
area = pi * radius ** 2
formated_string = 'The area of circle with a radius %d is %.2f.' %(radius, area)
print(formated_string) # 2 refers the 2 significant digits after the point

python_libraries = ['Django', 'Flask', 'NumPy', 'Matplotlib','Pandas']
formated_string = 'The following are python libraries:%s' % (python_libraries)
print(formated_string) # "The following are python libraries:['Django', 'Flask', 'NumPy', 'Matplotlib','Pandas']"
```

```python
# accessing the charector

language = 'Python'
first_letter = language[0]
print(first_letter) # P
second_letter = language[1]
print(second_letter) # y
last_index = len(language) - 1
last_letter = language[last_index]
print(last_letter) # n
```

```python
language = 'Python'
last_letter = language[-1]
print(last_letter) # n
second_last = language[-2]
print(second_last) # o
```

```python
name = "roshan Yadav"
print(name.count('r'))
print(name.capitalize())
print(name.endswith('n'))
string = "hello\tmy\tname\troshan"
print(string.expandtabs(2))
print(string.expandtabs(10))
print(name.find('n'))# return the index of first occurence
print(name.rfind('a'))# return the last occurence 

```

```python
# check for alpha numeric 
name = "30daysofpython"
name1 = "thirtydaysofpython"
name2 = "thirty day of python"# space are not allow
print(name.isalnum())
print(name1.isalnum())
print(name2.isalnum())
```

```python
challenge = 'thirty days of python'
print(challenge.isalpha()) # False, space is once again excluded
challenge = 'ThirtyDaysPython'
print(challenge.isalpha()) # True
num = '123'
print(num.isalpha())      # False
```
