```python
my_self = {
    'name':'Roshan Yadav',
    'class':'Btech RAI',
    'roll no':24001020028,
    'city':'faridabad',
    'skills':['python','cpp','java']
}
# adding the new key , value pair
my_self['intres']=['running']
print(type(my_self['intres']))
print(my_self)
print(len(my_self))
# when you ask for a key which are not available then it raised an error to prevent the erroe we use get method
print(my_self.get('hello'))
```

```python
person = {
    'first_name':'Roshan',
    'last_name':'Yadav',
    'age':21,
    'country':'India',
    'is_marred':False,
    'skills':['JavaScript', 'React', 'Node', 'MongoDB', 'Python'],
    'address':{
        'street':'Faridabad',
        'zipcode':'121001'
    }
    }
person['first_name'] = 'Roshu'
person['age'] = 252
print(person)

# you can modify the value 
```

```python
print('age' in person)
print('first_name'in person)
```

```python
# syntax
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
print(dct)
dct.pop('key1') # removes key1 item
print(dct)
dct = {'key1':'value1', 'key2':'value2', 'key3':'value3', 'key4':'value4'}
dct.popitem() # removes the last item
print(dct)
del dct['key2'] # removes key2 item

```

```python
# you can also convert this into list
print(dct.items())
print(type(dct.items()))
print(dct.values())
print(dct.keys())
```
